```python
from huggingface_hub import hf_hub_download
```

```python
import os
```

```python
os.makedirs("models", exist_ok=True)
```

```python
model_files = [
    "MedSAM2_2411.pt",
    "MedSAM2_US_Heart.pt",
    "MedSAM2_MRI_LiverLesion.pt",
    "MedSAM2_CTLesion.pt",
    "MedSAM2_latest.pt"
]
```

```python
for model_file in model_files:
    hf_hub_download(
        repo_id="wanglab/MedSAM2",
        filename=model_file,
        local_dir="models",
    )
    print(f"Downloaded {model_file}")
```
